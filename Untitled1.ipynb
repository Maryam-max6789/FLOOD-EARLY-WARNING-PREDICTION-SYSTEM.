```markdown
#  LSTM-Based Flood Prediction & River Level Forecasting

### Project Overview
This project implements a Deep Learning LSTM network to predict river water levels based on historical streamflow levels, upstream runoff, and precipitation indicators. This serves as a core engine for Flood Early Warning Systems (FEWS).

**Dataset :** For real-world deployment, this notebook can be trained directly on USGS streamflow datasets or regional rainfall-runoff Kaggle datasets.
```

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import math

# --- 1. SIMULATE RAINFALL & RIVER RUNOFF DATA ---
np.random.seed(101)
timestamps = pd.date_range(start="2021-01-01", periods=2000, freq='H')

# Generate simulated rainfall spikes (storm events)
rainfall = np.random.exponential(scale=0.5, size=2000)
rainfall[rainfall < 1.0] = 0.0  # Most hours have 0 rainfall

# Simulate river water level reacting to rainfall with a time lag
water_level = [1.5] # Base river level in meters
for i in range(1, 2000):
    # Water level decays slowly, but spikes when there is rainfall (with some routing delay)
    decay = (water_level[-1] - 1.5) * 0.95 + 1.5
    influence_from_rain = np.sum(rainfall[max(0, i-6):i]) * 0.4
    noise = np.random.normal(0, 0.05)
    water_level.append(decay + influence_from_rain + noise)

df_flood = pd.DataFrame({
    'Precipitation_mm': rainfall,
    'Water_Level_m': water_level
}, index=timestamps)

# --- 2. FEATURE ENGINEERING & LAGS ---
# Standard indicators for hydrological forecasting
df_flood['Precip_Cumulative_6H'] = df_flood['Precipitation_mm'].rolling(window=6).sum()
df_flood.fillna(0, inplace=True)

# Display simulated dataset snapshot
display(df_flood.head(10))
````

````python
# --- 3. DATA PREPARATION FOR SEQUENCE MODELING ---
features = ['Water_Level_m', 'Precipitation_mm', 'Precip_Cumulative_6H']
data_matrix = df_flood[features].values

# Train-Test Split (80% / 20% preserving time chronology)
train_len = int(len(data_matrix) * 0.8)
train_set = data_matrix[:train_len]
test_set = data_matrix[train_len:]

scaler_flood = MinMaxScaler(feature_range=(0, 1))
scaled_train_flood = scaler_flood.fit_transform(train_set)
scaled_test_flood = scaler_flood.transform(test_set)

# Convert to sequential blocks (Use past 24 hours to predict the next hour level)
def create_flood_sequences(data, lookback):
    X, y = [], []
    for i in range(len(data) - lookback):
        X.append(data[i:i+lookback])
        y.append(data[i+lookback, 0]) # Target: Next hour's Water_Level_m
    return np.array(X), np.array(y)

LOOKBACK_WINDOW = 24
X_train_f, y_train_f = create_flood_sequences(scaled_train_flood, LOOKBACK_WINDOW)
X_test_f, y_test_f = create_flood_sequences(scaled_test_flood, LOOKBACK_WINDOW)
````

````python
# --- 4. FLOOD LSTM ARCHITECTURE & TRAINING ---
flood_model = Sequential([
    LSTM(units=64, return_sequences=True, input_shape=(X_train_f.shape[1], X_train_f.shape[2])),
    Dropout(0.15),
    LSTM(units=32, return_sequences=False),
    Dropout(0.15),
    Dense(units=16, activation='relu'),
    Dense(units=1)
])

flood_model.compile(optimizer='adam', loss='mean_squared_error')

early_stop_f = EarlyStopping(monitor='val_loss', patience=4, restore_best_weights=True)
history_f = flood_model.fit(
    X_train_f, y_train_f,
    epochs=15,
    batch_size=32,
    validation_data=(X_test_f, y_test_f),
    callbacks=[early_stop_f],
    verbose=1
)
````

````python
# --- 5. INVERSE SCALING & EVALUATION ---
predictions_f = flood_model.predict(X_test_f)

# Scale predictions back to actual meters
pred_padded_f = np.zeros((len(predictions_f), len(features)))
pred_padded_f[:, 0] = predictions_f.flatten()
predictions_actual = scaler_flood.inverse_transform(pred_padded_f)[:, 0]

y_test_padded_f = np.zeros((len(y_test_f), len(features)))
y_test_padded_f[:, 0] = y_test_f
true_actual = scaler_flood.inverse_transform(y_test_padded_f)[:, 0]

# Calculate Metrical Performance
rmse_f = math.sqrt(mean_squared_error(true_actual, predictions_actual))
mae_f = mean_absolute_error(true_actual, predictions_actual)
r2_f = r2_score(true_actual, predictions_actual)

print(f"Flood Model Performance:")
print(f"- Root Mean Squared Error: {rmse_f:.3f} meters")
print(f"- Mean Absolute Error:      {mae_f:.3f} meters")
print(f"- Nash-Sutcliffe / R2 score: {r2_f:.4f}")
````

````python
# --- 6. VISUALIZATION OF HYDROLOGICAL FORECAST ---
plt.figure(figsize=(15, 6))
plt.plot(df_flood.index[train_len + LOOKBACK_WINDOW:], true_actual, label='Observed River Level', color='navy', lw=2)
plt.plot(df_flood.index[train_len + LOOKBACK_WINDOW:], predictions_actual, label='LSTM Forecast Level', color='crimson', linestyle='--', lw=2)
plt.axhline(y=2.5, color='orange', linestyle=':', label='Action Stage (Minor Flooding)')
plt.axhline(y=3.2, color='red', linestyle='-.', label='Flood Stage (Major Flooding)')
plt.title('River Stage Hydrograph: Observed vs LSTM Real-time Forecast', fontsize=16)
plt.xlabel('Timeline', fontsize=12)
plt.ylabel('River Height (meters)', fontsize=12)
plt.legend(loc='upper left')
plt.grid(True, alpha=0.3)
plt.show()
````

```markdown
## Part 8: Advanced Upgrade - Multi-Step Ahead Forecasting (6-Hour Horizon)

To make this a world-class portfolio project, we expand the LSTM model to perform **Multi-Step forecasting**. Instead of predicting just $t+1$, the model will simultaneously output predictions for the next 6 hours $[t+1, t+2, ..., t+6]$, allowing decision-makers to see the oncoming flood crest.
```

````python
# --- 8. MULTI-STEP AHEAD FORECASTING (6-HOUR WINDOW) ---
HORIZON = 6

def create_multistep_sequences(data, lookback, horizon):
    X, y = [], []
    for i in range(len(data) - lookback - horizon + 1):
        X.append(data[i:i+lookback])
        # Target is a vector of the next 'horizon' values of water level
        y.append(data[i+lookback : i+lookback+horizon, 0])
    return np.array(X), np.array(y)

X_train_ms, y_train_ms = create_multistep_sequences(scaled_train_flood, LOOKBACK_WINDOW, HORIZON)
X_test_ms, y_test_ms = create_multistep_sequences(scaled_test_flood, LOOKBACK_WINDOW, HORIZON)

# Build Multi-step LSTM Model
multistep_model = Sequential([
    LSTM(units=64, return_sequences=True, input_shape=(X_train_ms.shape[1], X_train_ms.shape[2])),
    Dropout(0.15),
    LSTM(units=32, return_sequences=False),
    Dropout(0.15),
    Dense(units=32, activation='relu'),
    Dense(units=HORIZON) # Output layer has nodes equivalent to target forecasting horizon
])

multistep_model.compile(optimizer='adam', loss='mean_squared_error')

# Fit the model quietly
history_ms = multistep_model.fit(
    X_train_ms, y_train_ms,
    epochs=15,
    batch_size=32,
    validation_data=(X_test_ms, y_test_ms),
    verbose=0
)

# Run prediction on the test set
predictions_ms = multistep_model.predict(X_test_ms)

# Visualize a specific sequence projection
sample_idx = 100
single_forecast_scaled = predictions_ms[sample_idx]
single_actual_scaled = y_test_ms[sample_idx]

padded_forecast = np.zeros((HORIZON, len(features)))
padded_forecast[:, 0] = single_forecast_scaled
forecast_meters = scaler_flood.inverse_transform(padded_forecast)[:, 0]

padded_actual = np.zeros((HORIZON, len(features)))
padded_actual[:, 0] = single_actual_scaled
actual_meters = scaler_flood.inverse_transform(padded_actual)[:, 0]

# Plot the comparative trajectory
plt.figure(figsize=(10, 5))
time_steps = np.arange(1, HORIZON + 1)
plt.plot(time_steps, actual_meters, marker='o', label='Observed (Actual) Path', color='navy', lw=2)
plt.plot(time_steps, forecast_meters, marker='x', linestyle='--', label='LSTM Multi-step Forecast', color='crimson', lw=2)
plt.title('6-Hour Ahead Future Water Level Forecast Trajectory', fontsize=14)
plt.xlabel('Hours into the Future (t + x)', fontsize=12)
plt.ylabel('River Height (meters)', fontsize=12)
plt.xticks(time_steps)
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
````

````python
# =====================================================================
# PORTFOLIO PROJECT: Multi-Feature Stock Price Forecasting using LSTM
# Recommended Kaggle Dataset: "Google Stock Price Prediction" or "Yahoo Finance"
# =====================================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

# --- 1. DATA ACQUISITION & SIMULATION ---
# To make this code instantly runnable, we generate realistic synthetic stock data.
# For your portfolio, replace this block by loading your downloaded Kaggle CSV:
# df = pd.read_csv('GOOG.csv', parse_dates=['Date'], index_index='Date')

np.random.seed(42)
dates = pd.date_range(start="2020-01-01", periods=1000, freq='B')
close_prices = 100 + np.cumsum(np.random.normal(0.1, 1.5, size=1000)) # Random walk
df = pd.DataFrame({'Close': close_prices}, index=dates)

# --- 2. FEATURE ENGINEERING ---
# Adding technical indicators to improve LSTM performance
df['MA_5'] = df['Close'].rolling(window=5).mean()
delta = df['Close'].diff()
up = delta.clip(lower=0)
down = -1 * delta.clip(upper=0)
ema_up = up.ewm(com=13, adjust=False).mean()
ema_down = down.ewm(com=13, adjust=False).mean()
rs = ema_up / ema_down
df['RSI'] = 100 - (100 / (1 + rs))

# Fill NaNs created by rolling windows
df.fillna(method='bfill', inplace=True)

# --- 3. DATA SPLITTING & SCALING ---
features = ['Close', 'MA_5', 'RSI']
data_filtered = df[features].values

train_size = int(len(data_filtered) * 0.8)
train_data = data_filtered[:train_size]
test_data = data_filtered[train_size:]

scaler = MinMaxScaler(feature_range=(0, 1))
scaled_train = scaler.fit_transform(train_data)
scaled_test = scaler.transform(test_data)

# --- 4. CREATE SEQUENCES FOR LSTM (3D Input: [samples, time steps, features]) ---
def create_sequences(data, seq_length):
    X, y = [], []
    for i in range(len(data) - seq_length):
        X.append(data[i:i+seq_length])
        y.append(data[i+seq_length, 0]) # Predicting the 'Close' price (index 0)
    return np.array(X), np.array(y)

SEQUENCE_LENGTH = 60
X_train, y_train = create_sequences(scaled_train, SEQUENCE_LENGTH)
X_test, y_test = create_sequences(scaled_test, SEQUENCE_LENGTH)

# --- 5. BUILD LSTM NETWORK ---
model = Sequential([
    LSTM(units=50, return_sequences=True, input_shape=(X_train.shape[1], X_train.shape[2])),
    Dropout(0.2),
    LSTM(units=50, return_sequences=False),
    Dropout(0.2),
    Dense(units=25),
    Dense(units=1)
])

model.compile(optimizer='adam', loss='mean_squared_error')
model.summary()

# --- 6. MODEL TRAINING ---
early_stop = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)
history = model.fit(
    X_train, y_train,
    epochs=20,
    batch_size=32,
    validation_data=(X_test, y_test),
    callbacks=[early_stop],
    verbose=1
)

# --- 7. PREDICTIONS & EVALUATION ---
predictions = model.predict(X_test)

# Pad predictions to inverse scale back to original dimensions
pred_padded = np.zeros((len(predictions), len(features)))
pred_padded[:, 0] = predictions.flatten()
predictions_transformed = scaler.inverse_transform(pred_padded)[:, 0]

# Pad true labels to inverse scale
y_test_padded = np.zeros((len(y_test), len(features)))
y_test_padded[:, 0] = y_test
y_test_transformed = scaler.inverse_transform(y_test_padded)[:, 0]

# --- 8. PORTFOLIO-READY VISUALIZATION ---
plt.figure(figsize=(14, 6))
plt.plot(df.index[train_size + SEQUENCE_LENGTH:], y_test_transformed, label='Actual Price', color='blue', lw=2)
plt.plot(df.index[train_size + SEQUENCE_LENGTH:], predictions_transformed, label='LSTM Predicted Price', color='orange', linestyle='--', lw=2)
plt.title('Portfolio Project: LSTM Stock Price Prediction Showcase', fontsize=16)
plt.xlabel('Date', fontsize=12)
plt.ylabel('Stock Price ($)', fontsize=12)
plt.legend(fontsize=12)
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
````
